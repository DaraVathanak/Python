# Automotive Trends - MPG Dataset
##### Dataset: sns.load_dataset('mpg') 
##### Goal: Compare car performance and fuel efficiency across different global regions.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df = sns.load_dataset('mpg')
df
```

##### FacetGrid / Relplot / Catplot: Use sns.relplot(x='weight', y='mpg', col='origin'). This will generate three separate scatterplots side-by-side (USA, Europe, Japan) so students can compare weight vs. fuel efficiency by region. Then, use sns.catplot(kind='box', x='cylinders', y='horsepower', row='origin') to create a vertical stack of boxplots.

```python
sns.relplot(data=df,x='weight', y='mpg', col='origin', kind='scatter')
plt.show()
```

##### This graph tells a simple story about cars from USA, Japan, and Europe.In all three groups, when a car becomes heavier, its mpg becomes lower. This means heavier cars usually use more fuel.USA cars are mostly heavier and have lower mpg.Japan cars are lighter and often have higher mpg.Europe cars are in the middle, but many light European cars also have good mpg.In simple words: lighter cars save more fuel, while heavier cars use more fuel.
##### ក្រាហ្វនេះប្រាប់អំពីឡានមកពី USA, Japan និង Europe។នៅក្នុងក្រុមទាំងបី ពេលឡានមាន ទម្ងន់ធ្ងន់ជាង, mpg ក៏ទាបជាង។ មានន័យថា ឡានធ្ងន់ប្រើប្រេងច្រើនជាង។ឡាន USA ភាគច្រើនធ្ងន់ ហើយ mpg ទាប។ឡាន Japan ស្រាលជាង ហើយ mpg ខ្ពស់ជាង។ឡាន Europe ស្ថិតនៅចន្លោះកណ្តាល ប៉ុន្តែឡានស្រាលៗក៏សន្សំសំចៃប្រេងបានល្អ។និយាយឱ្យងាយយល់៖ ឡានស្រាលសន្សំសំចៃប្រេងជាង ឡានធ្ងន់ប្រើប្រេងច្រើនជាង។

```python
sns.catplot(data=df,kind='box', x='cylinders', y='horsepower', row='origin')
plt.show()
```

##### This graph tells a story about car power in USA, Japan, and Europe.As the number of cylinders increases, the horsepower also increases.USA cars have the highest horsepower, especially 8-cylinder cars.Japan cars are mostly lower to medium horsepower.Europe cars are also mostly medium, but 6-cylinder cars are stronger.In simple words: more cylinders usually mean more horsepower, and USA cars are the most powerful in this graph.
##### ក្រាហ្វនេះប្រាប់អំពីកម្លាំងម៉ាស៊ីនឡាននៅ USA, Japan និង Europe។ពេលចំនួន cylinders កើនឡើង, horsepower ក៏កើនឡើងដែរ។ឡាន USA មានកម្លាំងខ្លាំងជាងគេ ជាពិសេសឡាន 8 cylinders។ឡាន Japan ភាគច្រើនមានកម្លាំងទាបទៅមធ្យម។ឡាន Europe ក៏ស្ថិតនៅកម្លាំងមធ្យម ប៉ុន្តែ 6 cylinders មានកម្លាំងខ្លាំងជាង។និយាយឱ្យងាយយល់៖ cylinders កាន់តែច្រើន horsepower ក៏កាន់តែខ្ពស់ ហើយឡាន USA ខ្លាំងជាងគេក្នុងក្រាហ្វនេះ។

##### Joint Plots: Use sns.jointplot(x='horsepower', y='mpg') to show a scatter plot in the center, with a histogram of horsepower on the top margin and a histogram of mpg on the right margin.

```python
sns.jointplot(data=df,x='horsepower', y='mpg',kind='scatter')
plt.show()
```

##### This graph tells a simple story about horsepower and fuel efficiency.When a car has low horsepower, it usually has higher mpg, meaning it saves more fuel.But when horsepower becomes higher, mpg goes down, meaning the car uses more fuel.In simple words: more power usually means less fuel efficiency.
##### ក្រាហ្វនេះប្រាប់អំពី horsepower និងការសន្សំសំចៃប្រេង។ពេលឡានមាន horsepower ទាប, វាច្រើនមាន mpg ខ្ពស់ មានន័យថាសន្សំសំចៃប្រេងជាង។ប៉ុន្តែពេល horsepower ខ្ពស់ឡើង, mpg ក៏ធ្លាក់ចុះ មានន័យថាប្រើប្រេងច្រើនជាង។និយាយឱ្យងាយយល់៖ ឡានកម្លាំងខ្លាំង ច្រើនតែស៊ីប្រេងជាង។
