# Restaurant Economics
##### Dataset: sns.load_dataset('tips') 

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
df = sns.load_dataset('tips')
df = df.reset_index()
df
```

### Goal: Find out what factors drive higher tips at a restaurant.
##### Num vs. Num - Scatter: Create a sns.scatterplot() with total_bill on the X-axis and tip on the Y-axis to prove that higher bills equal higher tips.

```python
sns.scatterplot(x='total_bill', y='tip', data=df)
plt.show()
```

##### This scatter plot tells a simple story: customers who spend more usually leave higher tips. Most small bills receive smaller tips, while larger bills often lead to bigger tips. However, the points are not perfectly lined up, which means bill amount is not the only reason people tip more. Some customers with high bills still leave average tips, while others are more generous. Overall, total_bill is a strong factor, but other things like service, party size, and dining time may also affect the tip.
##### Scatter Plot នេះប្រាប់រឿងមួយថា អតិថិជនដែលចំណាយច្រើន ជាទូទៅឱ្យ Tip ច្រើនជាង។ វិក្កយបត្រតូចៗភាគច្រើនទទួលបាន Tip តិច ខណៈវិក្កយបត្រធំៗភាគច្រើនទទួលបាន Tip ខ្ពស់។ ប៉ុន្តែចំណុចទាំងអស់មិនស្ថិតនៅលើបន្ទាត់តែមួយទេ មានន័យថា total_bill មិនមែនជាកត្តាតែមួយទេ។ សរុបមក វិក្កយបត្រសរុបជាកត្តាសំខាន់ ប៉ុន្តែសេវាកម្ម ចំនួនមនុស្ស និងពេលវេលាអាហារ ក៏អាចប៉ះពាល់ដល់ Tip ផងដែរ។

##### Num vs. Cat - Box Plots: Use sns.boxplot() with day on the X-axis and total_bill on the Y-axis to find the median bill and outliers for each day of the week.

```python
sns.boxplot(x='day', y='total_bill', data=df)
plt.show()
```

##### This box plot tells a clear story: customers spend more on weekends than on weekdays. Thursday and Friday have lower median bills, while Saturday and Sunday show higher typical spending. The weekend days also have wider bill ranges, meaning customer spending is more varied. The outliers show a few very high bills, especially on Saturday and Sunday, suggesting that weekends bring larger groups or more expensive orders.
##### Box Plot នេះបង្ហាញថា អតិថិជនចំណាយច្រើនជាងនៅថ្ងៃចុងសប្តាហ៍។ Thursday និង Friday មានវិក្កយបត្រកណ្ដាលទាបជាង ខណៈ Saturday និង Sunday មានការចំណាយជាមធ្យមខ្ពស់ជាង។ ថ្ងៃចុងសប្តាហ៍ក៏មានការប្រែប្រួលតម្លៃច្រើនជាងដែរ។ ចំណុច Outliers បង្ហាញថាមានវិក្កយបត្រខ្ពស់ខ្លាំងមួយចំនួន ជាពិសេសនៅ Saturday និង Sunday ដែលអាចបង្ហាញពីក្រុមអតិថិជនធំ ឬការបញ្ជាទិញថ្លៃជាង។

##### Num vs. Cat - Violin Plots: Use sns.violinplot() comparing time (Lunch vs. Dinner) and tip to see where the density of generous tippers lies.

```python
sns.violinplot(x='time', y='tip', data=df)
plt.show()
```

##### This violin plot tells a simple story: dinner customers tend to leave higher and more varied tips than lunch customers. Lunch tips are mostly concentrated around 2USD, showing a more predictable tipping pattern. Dinner, however, has a wider spread, with many tips around 2–4USD and a few very generous tips reaching above 6USD and even close to 10USD. This suggests that dinner service may bring larger bills, bigger groups, or more generous customers.
##### Violin Plot នេះបង្ហាញថា អតិថិជនពេល Dinner ជាទូទៅឱ្យ Tip ខ្ពស់ និងមានការប្រែប្រួលច្រើនជាងពេល Lunch។ Tip ពេល Lunch ភាគច្រើនប្រមូលផ្តុំនៅជុំវិញ 2USD ដែលបង្ហាញថាការឱ្យ Tip មានលក្ខណៈថេរជាង។ ចំណែក Dinner មានការបែងចែកធំជាង ដោយមាន Tip ច្រើននៅចន្លោះ 2–4USD និងមាន Tip ខ្ពស់ៗខ្លះលើស 6USD រហូតជិត 10USD។ នេះបង្ហាញថាពេល Dinner អាចមានវិក្កយបត្រធំជាង ក្រុមអតិថិជនច្រើនជាង ឬអតិថិជនមានចិត្តទូលាយជាង។

##### Num vs. Cat - Bar Plots: Use sns.barplot() to show the average tip amount grouped by size (party size), letting Seaborn calculate the mean automatically.

```python
sns.barplot(x='size', y='tip', estimator='mean', data=df)
plt.show()
```

##### This bar plot tells a clear story: larger groups generally leave higher tips. Party sizes of 1 and 2 give smaller average tips, while groups of 4, 5, and especially 6 people leave much larger tips. This suggests that bigger tables usually create higher bills, which leads to higher tip amounts. However, the error bars show some variation, especially for larger groups, meaning not every big group tips the same way.
##### Bar Plot នេះបង្ហាញថា ក្រុមអតិថិជនដែលមានមនុស្សច្រើន ជាទូទៅឱ្យ Tip ច្រើនជាង។ ក្រុមដែលមាន 1 ឬ 2 នាក់ ឱ្យ Tip ជាមធ្យមតិចជាង ខណៈក្រុម 4, 5 និងជាពិសេស 6 នាក់ ឱ្យ Tip ខ្ពស់ជាង។ នេះបង្ហាញថាតុធំៗភាគច្រើនមានវិក្កយបត្រខ្ពស់ ហើយនាំឱ្យ Tip ខ្ពស់។ ប៉ុន្តែ Error Bars បង្ហាញថា មានការប្រែប្រួលខ្លះ ជាពិសេសសម្រាប់ក្រុមធំៗ។
