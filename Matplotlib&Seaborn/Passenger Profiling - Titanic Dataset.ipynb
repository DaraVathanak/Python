# Passenger Profiling - Titanic Dataset
##### Dataset: sns.load_dataset('titanic') 

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
df = sns.load_dataset('titanic') 
df
```

### Goal: Understand who was on the Titanic by exploring single variables.
##### Categorical - Bar/Count: Use sns.countplot() to show how many passengers were in First, Second, and Third class (class column).

```python
sns.countplot(data=df, x='class')
plt.title('Passenger class as a word')
plt.show()
```

##### In total, there are about 891 passengers shown in the graph. Third Class is the most common group, with about 491 passengers. First Class has about 216 passengers, and Second Class has the fewest, with about 184 passengers. Overall, most passengers are in Third Class, while First and Second Class have much smaller numbers.
##### សរុបមកមានអ្នកដំណើរប្រហែល 891 នាក់ដែលបង្ហាញក្នុងក្រាហ្វនេះ។ Third Class ជាក្រុមដែលមានចំនួនច្រើនជាងគេដោយមានប្រហែល 491 នាក់។ First Class មានប្រហែល 216 នាក់ហើយ Second Class មានតិចជាងគេប្រហែល 184 នាក់។ សរុបមកអ្នកដំណើរភាគច្រើនស្ថិតនៅក្នុង Third Class ខណៈដែល First និង Second Class មានចំនួនតិចជាង។

##### Categorical - Pie: Use plt.pie() to show the raw proportion of male vs. female passengers (sex column), discussing why the countplot might have been easier to read.

```python
count = df['sex'].value_counts()
plt.pie(count, labels=count.index, autopct='%1.1f%%')
plt.title('Percentage of Male and Female')
plt.show()
```

##### Numerical - KDE: Use sns.kdeplot() to look at the density of ticket prices (fare column) to see the massive skew toward cheaper tickets.

```python
sns.kdeplot(data=df, x='fare', fill=True)
plt.title('Density of ticket prices')
plt.show()
```

##### The giant hill on the left side of the graph tells us that a massive group of passengers paid very similar low ticket prices, clustering near the cheaper fare range. But as you move further to the right and look at higher ticket prices like 100, 200, or even over 500, the number of passengers becomes very small very quickly. That long tail stretching out to the right shows that while a few rare passengers paid very expensive fares, the vast majority of passengers were gathered together in that big peak on the lower-price side.
##### កំពូលធំនៅខាងឆ្វេងនៃក្រាហ្វនេះបង្ហាញថាអ្នកដំណើរភាគច្រើនបានបង់ថ្លៃសំបុត្រទាបដែលស្ថិតនៅជិតៗគ្នា។ ប៉ុន្តែពេលមើលទៅខាងស្តាំដែលជាតម្លៃសំបុត្រថ្លៃៗដូចជា 100, 200 ឬលើស 500 ចំនួនអ្នកដំណើរធ្លាក់ចុះយ៉ាងលឿន។ ខ្សែវែងដែលលាតទៅខាងស្តាំបង្ហាញថាទោះបីមានអ្នកដំណើរមួយចំនួនតូចបានបង់ថ្លៃសំបុត្រថ្លៃខ្លាំងក៏ដោយអ្នកដំណើរភាគច្រើនស្ថិតនៅក្រុមតម្លៃសំបុត្រថោកជាង។

##### Combining Hist + KDE: Overlay the KDE curve on top of a histogram for the age column using sns.histplot(data, x='age', kde=True) to see both the bins and the smooth trend.

```python
sns.histplot(data=df, x='age', bins=16, kde=True)
plt.title('Age Distribution with KDE')
plt.show()
```

##### This chart shows the age distribution with KDE, and it tells a clear story: most passengers are younger adults. The tallest bars are around the 20 to 30 age range, meaning this is where the largest group of passengers is found. Layered over the bars is a smooth KDE line, which helps show the overall shape of the age distribution more clearly. The smooth curve rises quickly from the younger ages and reaches its highest point around the mid-to-late 20s, showing that this is the main center of the passenger group. After that, the curve slowly moves downward. Once you look past age 40, the number of passengers starts to decline more steadily. By the time the ages reach the 60s, 70s, and 80s, there are only a small number of passengers left. Overall, this chart tells us that the passengers are mostly young and middle-aged adults, while older passengers are much less common.
